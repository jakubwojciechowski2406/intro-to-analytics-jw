**A. LSTM: long-Short Term Memory Framework**

**A.I Training**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import KFold

# ===============================
# PARAMETERS
# ===============================
SEQ_LEN = 20  # Number of past days to look at
EPOCHS = 50
BATCH_SIZE = 16
FOLDS = 3

'''The above parameters are subject to change. So change them for experimental purpose'''

# ===============================
# DATA PREPARATION
# ===============================
# Use non-differenced Close prices
series = df['Close'].values.reshape(-1, 1)

# Scale data
scaler = MinMaxScaler()
series_scaled = scaler.fit_transform(series)

# Create sequences
def create_sequences(data, seq_len):
    X, y = [], []
    for i in range(len(data) - seq_len):
        X.append(data[i:i+seq_len])
        y.append(data[i+seq_len])
    return np.array(X), np.array(y)

X, y = create_sequences(series_scaled, SEQ_LEN)
print(f"Sequences created: X={X.shape}, y={y.shape}")

# ===============================
# K-FOLD CROSS-VALIDATION
# ===============================
kf = KFold(n_splits=FOLDS, shuffle=False)  # time series: do not shuffle for real CV

fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
    print(f"\n=== Fold {fold+1} ===")
    X_train, X_val = X[train_idx], X[val_idx]
    y_train, y_val = y[train_idx], y[val_idx]

    # Build LSTM model
    model = Sequential([
        LSTM(64, input_shape=(SEQ_LEN, 1), return_sequences=False),
        Dropout(0.2),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')

    # Early stopping
    es = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True, verbose=1)

    # Train model
    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=[es],
        verbose=1
    )

    # Predictions
    y_pred = model.predict(X_val)

    # Inverse scaling
    y_val_inv = scaler.inverse_transform(y_val)
    y_pred_inv = scaler.inverse_transform(y_pred)

    # Metrics
    rmse = np.sqrt(mean_squared_error(y_val_inv, y_pred_inv))
    mae = mean_absolute_error(y_val_inv, y_pred_inv)
    mape = np.mean(np.abs((y_val_inv - y_pred_inv) / y_val_inv)) * 100

    fold_metrics.append({'fold': fold+1, 'rmse': rmse, 'mae': mae, 'mape': mape})
    print(f"Fold {fold+1} Metrics -> RMSE: {rmse:.2f}, MAE: {mae:.2f}, MAPE: {mape:.2f}%")

    # ===============================
    # Visualization
    # ===============================
    plt.figure(figsize=(12,4))
    plt.plot(y_val_inv, label='Actual', color='black')
    plt.plot(y_pred_inv, label='Predicted', color='red', alpha=0.7)
    plt.title(f'Fold {fold+1} LSTM Predictions')
    plt.xlabel('Time Step')
    plt.ylabel('Price')
    plt.legend()
    plt.show()

# ===============================
# SUMMARY METRICS
# ===============================
metrics_df = pd.DataFrame(fold_metrics)
print("\n=== Cross-Validation Metrics Summary ===")
print(metrics_df)
print("\nAverage Metrics:")
print(metrics_df.mean())
```

**How the code works:**
1. **Parameters:** `SEQ_LEN = 20` – the model looks at the last 20 days to predict the next day. Max 50 epochs, batch size 16, 3 folds.
2. **Scaling:** `MinMaxScaler` changes prices to the range 0–1, which makes neural network training easier.
3. **Sequences:** `create_sequences` builds pairs – X = 20 past prices, y = the price on the next day.
4. **Cross-validation:** `KFold` with 3 folds, no shuffle. In each fold a new model is trained on 2 parts and tested on the third part.
5. **Model:** LSTM layer with 64 units (it keeps memory of the sequence) → Dropout 0.2 (against overfitting) → Dense(1) (one output = next-day price). Optimizer Adam, loss MSE.
6. **EarlyStopping:** stops when validation loss does not improve for 5 epochs and restores the best weights.
7. **Metrics:** predictions are scaled back to USD and compared with real prices (RMSE, MAE, MAPE).

**Change vs the original:** `SEQ_LEN` was 1 here but 20 in the forecasting cell. Now it is 20 in both, so the forecast uses the same input as in training.

**Results:**

| Fold | Period | RMSE | MAE | MAPE |
|---|---|---|---|---|
| 1 | 2010–2015 | 0.53 USD | 0.41 USD | 2.45% |
| 2 | 2015–2021 | 2.28 USD | 1.46 USD | 2.61% |
| 3 | 2021–2026 | 6.23 USD | 4.68 USD | 2.42% |
| Average | | 3.01 USD | 2.18 USD | 2.49% |

The predictions follow the real price very closely in all folds. RMSE in USD grows from fold 1 to 3 only because the prices are higher, the % error (MAPE) stays at about 2.5%. The prediction is close to "tomorrow ≈ today", which is typical for stock prices.

**Limitations:** `KFold` is not a real time-series split – in folds 1 and 2 the model is trained (partly) on later data and tested on earlier data. `TimeSeriesSplit` would be more correct. The scaler is fitted on all data, including the validation parts (small data leakage).

```python
model = Sequential([
    LSTM(64, input_shape=(SEQ_LEN, 1)),
    Dropout(0.2),
    Dense(1)
])
model.compile(optimizer='adam', loss='mse')
model.fit(X, y, epochs=20, batch_size=BATCH_SIZE, verbose=1)
```

**Final model:** after the loop, `model` was only the model from the last fold (trained on 2010–2021 data, prices up to 139 USD). For the forecast we train a final LSTM on all data, 20 epochs (final loss about 0.0004 on scaled data).

**2. Transformers**

In following we are leveraging the tranformers

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, LayerNormalization, Dropout, MultiHeadAttention, Add, GlobalAveragePooling1D
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import KFold
import tensorflow as tf

# ===============================
# PARAMETERS
# ===============================
SEQ_LEN = 30
EPOCHS = 50
BATCH_SIZE = 16
FOLDS = 5
HEADS = 4
FF_DIM = 32
DROPOUT = 0.1

# ===============================
# DATA PREPARATION
# ===============================
series = df['Close'].values.reshape(-1,1)
scaler = MinMaxScaler()
series_scaled = scaler.fit_transform(series)

def create_sequences(data, seq_len):
    X, y = [], []
    for i in range(len(data) - seq_len):
        X.append(data[i:i+seq_len])
        y.append(data[i+seq_len])
    return np.array(X), np.array(y)

X, y = create_sequences(series_scaled, SEQ_LEN)
print(f"X shape: {X.shape}, y shape: {y.shape}")

# ===============================
# HELPER FUNCTIONS
# ===============================
def transformer_encoder(inputs, head_size, ff_dim, dropout=0):
    # Multi-Head Self Attention
    x = MultiHeadAttention(num_heads=head_size, key_dim=ff_dim, dropout=dropout)(inputs, inputs)
    x = Dropout(dropout)(x)
    x = Add()([x, inputs])
    x = LayerNormalization(epsilon=1e-6)(x)

    # Feed-forward
    x_ff = Dense(ff_dim, activation='relu')(x)
    x_ff = Dense(inputs.shape[-1])(x_ff)
    x_ff = Dropout(dropout)(x_ff)
    x = Add()([x, x_ff])
    x = LayerNormalization(epsilon=1e-6)(x)
    return x

def calculate_metrics(actual, pred):
    rmse = np.sqrt(mean_squared_error(actual, pred))
    mae = mean_absolute_error(actual, pred)
    mape = np.mean(np.abs((actual - pred) / actual)) * 100
    return rmse, mae, mape

# ===============================
# K-FOLD CROSS VALIDATION
# ===============================
kf = KFold(n_splits=FOLDS, shuffle=False)
fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
    print(f"\n=== Fold {fold+1} ===")
    X_train, X_val = X[train_idx], X[val_idx]
    y_train, y_val = y[train_idx], y[val_idx]

    # Build Transformer model
    inputs = Input(shape=(SEQ_LEN,1))
    x = Dense(32)(inputs)
    x = transformer_encoder(x, HEADS, FF_DIM, DROPOUT)
    x = GlobalAveragePooling1D()(x)
    outputs = Dense(1)(x)
    model = Model(inputs, outputs)
    model.compile(optimizer='adam', loss='mse')

    # Early stopping
    es = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True, verbose=1)

    # Train model
    history = model.fit(X_train, y_train,
                        validation_data=(X_val, y_val),
                        epochs=EPOCHS,
                        batch_size=BATCH_SIZE,
                        callbacks=[es],
                        verbose=1)

    # Predict
    y_pred = model.predict(X_val)
    y_val_inv = scaler.inverse_transform(y_val)
    y_pred_inv = scaler.inverse_transform(y_pred)

    # Metrics
    rmse, mae, mape = calculate_metrics(y_val_inv, y_pred_inv)
    fold_metrics.append({'fold': fold+1, 'rmse': rmse, 'mae': mae, 'mape': mape})
    print(f"Fold {fold+1} Metrics -> RMSE: {rmse:.2f}, MAE: {mae:.2f}, MAPE: {mape:.2f}%")

    # Visualization
    plt.figure(figsize=(12,4))
    plt.plot(y_val_inv, label='Actual', color='black')
    plt.plot(y_pred_inv, label='Predicted', color='red', alpha=0.7)
    plt.title(f'Transformer Fold {fold+1} Predictions')
    plt.xlabel('Time Step')
    plt.ylabel('Price')
    plt.legend()
    plt.show()

# Summary
metrics_df = pd.DataFrame(fold_metrics)
print("\n=== Cross-Validation Metrics Summary ===")
print(metrics_df)
print("\nAverage Metrics:")
print(metrics_df.mean())
```

**How the code works:**
1. **Parameters:** window of 30 days, 5 folds, 4 attention heads, feed-forward size 32, dropout 0.1.
2. **Data:** the same scaling and sequences as for LSTM, but with 30 days.
3. **`transformer_encoder`:**
   - **MultiHeadAttention** (self-attention) – every day in the window looks at all other days and learns which ones are important.
   - **Add + LayerNormalization** – residual connection and normalization, it keeps training stable.
   - **Feed-forward** – two Dense layers applied to each day.
4. **Model:** input → Dense(32) → encoder → GlobalAveragePooling1D (average over the 30 days) → Dense(1) = next-day price.
5. **Training and metrics:** the same as for LSTM (Adam, MSE, EarlyStopping, RMSE / MAE / MAPE).

**Changes vs the original:**
- **Bug fix – `Dense(32)` before the encoder.** In the original code the input had only 1 feature per day. LayerNormalization normalizes over the features, and with 1 feature the result is always the same constant, so all information from the data was lost. The model could only give a constant (a flat line in every fold). The Dense(32) layer gives 32 features per day, so the normalization works.
- `FOLDS` 10 → 5 (faster training) and EarlyStopping `patience` 5 → 10 (before, training stopped too early – the best epoch was often 1 or 2).

**Results:**

| Fold | Period | RMSE | MAE | MAPE |
|---|---|---|---|---|
| 1 | 2010–2013 | 2.32 USD | 1.94 USD | 16.68% |
| 2 | 2013–2017 | 1.52 USD | 1.19 USD | 5.15% |
| 3 | 2017–2020 | 3.46 USD | 2.41 USD | 5.36% |
| 4 | 2020–2023 | 8.05 USD | 6.57 USD | 5.00% |
| 5 | 2023–2026 | 29.29 USD | 22.44 USD | 9.41% |
| Average | | 8.93 USD | 6.91 USD | 8.32% |

Now the model learns: in folds 2–4 the prediction follows the price (smoother and a bit delayed). Fold 1 has a high MAPE because the prediction is far too low at the start and prices were only 7–21 USD there. Fold 5 fails: in training the model never saw prices above 191 USD, so it cannot predict the real 163–341 USD range (the prediction stays at about 200–255 USD). The Transformer is clearly worse than LSTM (average MAPE 8.32% vs 2.49%).
