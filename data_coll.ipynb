```python
import yfinance as yf
import pandas as pd
```

```python
# Define ticker symbol for Google (Alphabet)
ticker = "GOOGL"

# Create a Ticker object
goog = yf.Ticker(ticker)
```

```python
data = goog.history(start="2026-01-01", end="2026-09-28", interval="1d")
```

```python
# Print the first few rows
print(data.head())
```

```python
# Extract only the closing price column
close_prices = data["Close"]

# If you want, convert to a DataFrame with date as a column
df_close = pd.DataFrame({"Date": close_prices.index, "Close": close_prices.values})
df_close.reset_index(drop=True, inplace=True)

print(df_close.head())

df_close.to_csv("google_close_prices.csv", index=False)
```

```python
ticker = 'AAPL'
aapl = yf.Ticker(ticker)
data_aapl = aapl.history(start="2026-01-01", end="2026-09-28", interval="1d")
print(data_aapl.head())
```

```python
close_prices_aapl = data_aapl["Close"]

df_close_aapl = pd.DataFrame({"Date": close_prices_aapl.index, "Close": close_prices_aapl.values})
df_close_aapl.reset_index(drop=True, inplace=True)

print(df_close_aapl.head())

df_close_aapl.to_csv("apple_close_prices.csv", index=False)
```
